### 一、venv (内置模块)

### 1、创建虚拟环境
#### 在当前目录下创建名为 .venv 的虚拟环境（推荐命名）
python3 -m venv .venv  或 python -m venv .venv
#### 创建自定义名称的环境
python3 -m venv myenv
#### 在指定路径创建环境
python3 -m venv ~/projects/myapp/venv
#### 不带 pip 的环境（极少使用）
python3 -m venv --without-pip .venv
#### 清空已有的环境目录（如果存在）
python3 -m venv --clear .venv
#### 使用系统 site-packages（不隔离系统包）
python3 -m venv --system-site-packages .venv
#### 升级环境到当前 Python 版本
python3 -m venv --upgrade .venv
#### 创建时指定 Python 版本（需要多个 Python 版本）
python3.11 -m venv .venv

### 2. 激活虚拟环境
#### Windows下：
CMD<br>
.venv\Scripts\activate.bat
<br>PowerShell <br>
.venv\Scripts\Activate.ps1
<br>Git Bash <br>
source .venv/Scripts/activate

#### Linux/macOS下：
bash/zsh<br>
source .venv/bin/activate
<br>fish shell<br>
source .venv/bin/activate.fish
<br>csh/tcsh<br>
source .venv/bin/activate.csh

### 3. 验证激活成功
终端提示符前会显示环境名称<br>
(.venv) user@host:~/project$

查看当前 Python 路径<br>
which python    # Linux/macOS<br>
where python    # Windows<br>

查看 Python 版本<br>
python --version

### 4. 虚拟环境内的工作命令
安装包<br>
pip install <包名><br>
pip install requests==2.28.1  # 指定版本<br>
pip install requests django flask  # 安装多个<br>

查看已安装的包<br>
pip list<br>
pip list --format=freeze  # 生成 requirements 格式  <br>

导出依赖到文件<br>
pip freeze > requirements.txt <br>

从文件安装依赖<br>
pip install -r requirements.txt<br>

升级包<br>
pip install --upgrade <包名><br>

卸载包<br>
pip uninstall <包名><br>

查看包详细信息<br>
pip show <包名><br>

检查依赖冲突<br>
pip check<br>

查看 Python 解释器路径<br>
python -c "import sys; print(sys.executable)"<br>

查看 site-packages 路径<br>
python -c "import site; print(site.getsitepackages())"<br>

查看环境变量<br>
echo $VIRTUAL_ENV  # Linux/macOS<br>
echo %VIRTUAL_ENV%  # Windows<br>

### 5.退出虚拟环境---退出后，终端提示符恢复正常，python 将指向系统 Python。
直接输入命令<br>
deactivate<br>

或者（不常用）<br>
source .venv/bin/deactivate  # Linux/macOS<br>

### 6. 删除虚拟环境
Linux/macOS下：<br>
rm -rf .venv<br>

Windows下：<br>
rmdir /s .venv<br>
或使用 PowerShell<br>
Remove-Item -Recurse -Force .venv<br>

### 7.修改默认的 pip 源（加速下载）
在虚拟环境中配置清华源<br>
pip config set global.index-url https://pypi.tuna.tsinghua.edu.cn/simple
