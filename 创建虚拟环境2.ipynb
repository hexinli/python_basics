### Miniconda 虚拟环境命令详解
Miniconda 是 Anaconda 的轻量级版本，只包含 conda 包管理器和 Python 环境，让你可以按需安装需要的包，避免占用过多磁盘空间。

### 1、创建虚拟环境
基础语法<br>
conda create -n <环境名> [python=<版本>] [包名列表]<br>
创建最小环境（只有 Python）<br>
	conda create -n myenv <br>
指定 Python 版本（推荐）<br>
    conda create -n myenv python=3.9<br>
创建时同时安装包<br>
    conda create -n myenv python=3.9 numpy pandas<br>
安装特定版本的包<br>
	conda create -n myenv python=3.9 pandas=1.3.0<br>
使用 --yes 自动确认<br>
conda create --yes -n myenv python=3.9<br>
示例：创建一个名为 "nlp_project" 的环境，Python 3.10，包含 numpy 和 pandas<br>
conda create -n nlp_project python=3.10 numpy pandas
#### 指定路径创建（高级用法）
默认情况下，环境创建在 ~/miniconda3/envs/ 目录下。如果需要将环境放在其他位置（如项目目录或共享存储），可以使用 --prefix：
##### 在指定目录创建环境
conda create --prefix /path/to/project/env python=3.9
##### 激活时需要指定完整路径
conda activate /path/to/project/env
##### 提示：使用 --prefix 创建的环境管理起来不如 -n 方便，一般推荐使用 -n 让 conda 统一管理。

### 2. 激活与退出虚拟环境
##### A、激活环境
conda activate <环境名>
##### B、查看当前激活的环境
终端提示符会显示当前环境名<br>
(myenv) user@host:~$

或使用命令查看<br>
conda info --envs  # 当前激活的环境前会有 * 号
##### C、退出环境
conda deactivate

### 3. 查看所有环境
##### 方式一

##### 方式二

### 4. 环境内包管理
##### 在激活的环境下操作
##### A、激活环境
conda activate myenv

##### B、安装包
conda install numpy <br>
conda install pandas=1.5.0        # 指定版本 <br>
conda install -c conda-forge scipy # 从特定频道安装[citation:2] <br>
##### C、查看已安装的包
conda list
##### D、更新包
conda update numpy
##### E、卸载包
conda remove numpy
#### 在指定环境中操作（无需激活）
##### A、安装包到指定环境
conda install -n myenv numpy
##### B、查看指定环境的包列表
conda list -n myenv
##### C、更新指定环境的包
conda update -n myenv numpy
##### D、卸载指定环境的包
conda remove -n myenv numpy

### 5. 复制与导出环境
#### 复制环境
##### 完整复制一个已有环境，conda 不支持直接重命名环境，如果需要重命名，可以通过 clone 后删除原环境的方式实现
conda create -n new_env --clone old_env
#### 导出环境配置
导出当前环境的完整依赖（包含 conda 包）<br>
conda env export > environment.yml<br>
仅导出显式安装的包（更推荐，跨平台兼容性更好）<br>
conda env export --from-history > environment.yml<br>
#### 从配置文件创建环境
conda env create -f environment.yml

### 6. 删除虚拟环境
##### 删除指定环境（注意：不可恢复）
conda remove -n myenv --all
* 注意：删除环境前请确保已退出该环境（conda deactivate）

### 8. 实用技巧
#### 配置国内镜像源（加速下载）
添加清华源<br>
conda config --add channels https://mirrors.tuna.tsinghua.edu.cn/anaconda/pkgs/main/<br>
conda config --add channels https://mirrors.tuna.tsinghua.edu.cn/anaconda/pkgs/free/<br>
conda config --add channels https://mirrors.tuna.tsinghua.edu.cn/anaconda/cloud/conda-forge/<br>

设置显示通道地址<br>
conda config --set show_channel_urls yes
