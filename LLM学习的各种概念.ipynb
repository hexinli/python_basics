## 养龙虾5 大核心术语

### 1. LLM（大语言模型）—— “龙虾的大脑”
这是AI的核心思考器官，也就是常说的“大模型”。

* 通俗解释：你可以把它理解成一个读过全世界所有公开书籍、文章、代码的超级大学霸。你问他问题，他能根据学过的知识生成回答。

* 在“养虾”中的角色：LLM就是“龙虾”的大脑。像GPT、Claude、文心一言这些，都是不同的“大脑”品牌。你“养虾”时，必须给它接上一个LLM大脑，它才会思考。大脑越聪明、知识越新，你的“虾”就越能干。

### 2. Agent（智能体）—— “龙虾的整个身体”
这是最近最核心的概念。

* 通俗解释：LLM只是大脑，只会想不会做。Agent就是一个配备了手脚和工具的完整机器人。它会自己规划步骤、使用工具、反思错误，去完成一个目标。

* 在“养虾”中的角色：OpenClaw（你养的龙虾）本身就是一个Agent。 比如你让它“写个游戏市场分析报告”，它作为Agent，会自己拆解任务：先去联网搜索最新数据，再打开Excel分析，最后打开Word把报告写好保存。整个过程自己规划、自己完成。

### 3. Skill（技能）—— “龙虾的可用工具”
这个很好懂。

* 通俗解释：就是Agent“手脚”能做的具体动作，一项Skill就是一个工具。比如“计算器”Skill能做数学题，“联网搜索”Skill能查最新信息。

* 在“养虾”中的角色：你给“龙虾”安装不同的Skill，它就学会了不同的本事。一个Skill就像手机上的一个App。 装个“邮件Skill”就能收发邮件，装个“浏览器Skill”就能上网。你的“龙虾”到底有多大能耐，全看给它装了多少Skill。

### 4. MCP（模型上下文协议）—— “龙虾的USB接口”
这是一个技术标准，但比喻一下也很好理解。

* 通俗解释：你可以把它想象成电脑上的USB接口标准。无论是U盘、鼠标还是键盘，只要有USB接口，插上就能用。MCP就是AI界的USB标准接口。

* 在“养虾”中的角色：MCP让各种工具和AI大脑之间，有了一个通用的连接方式。任何一个Skill如果遵循MCP标准，就像USB设备一样，可以被你的“龙虾”快速识别和调用，不需要复杂的定制开发。有了MCP，AI连接工具就像用USB一样即插即用。

### 5. Token（令牌/词元）—— “龙虾的算钱单位”
这是最重要、也最容易花钱的词。

* 通俗解释：这是LLM处理文本的最小单位，也是收费的基本单位。你可以把它理解为AI大脑的“脑细胞”。AI思考时，每处理或生成一个词，就要消耗掉一些“脑细胞”。

* 在“养虾”中的角色：这是你养虾的直接成本。一个中文字大概消耗1.5-2个Token。你问AI一句话，它会消耗输入（你的问题）和输出（AI的回答）两部分的Token。你花掉的每一分钱，本质上都是在为消耗的Token买单。

##### 四、一句话记住它们
###### 概念	一句话记忆
* LLM	只动脑，不动手（思考、推理、生成话术）
* Tool	一个螺丝刀（单一动作：查天气、搜票）
* Skill	老师的教案（把多个Tool串起来，带判断逻辑）
* MCP	万能插头（统一标准，换工具不换代码）
* Agent	干活的项目经理（拆任务、调人、处理异常、最终交付）
###### 现在再看这些概念，应该不再是孤立的名词，而是一个各司其职的团队了。
